What is Stemming?

Stemming is a text preprocessing technique in Natural Language Processing (NLP) 
that reduces a word to its root form (stem) by removing prefixes or suffixes.

The stem is not always a valid English word; it is simply a shortened form used to group similar words together.

Types:

PorterStemmer

SnowballStemmer

LancasterStemmer

RegexpStemmer


```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = ["playing", "played", "plays", "studies", "studying","eating","eat","eaten","goes","go","female","male","history"]

for word in words:
    print(f"{word} --> {stemmer.stem(word)}")
```

```python
#Disadvantages of PorterStemmer
# Can produce non-dictionary words (e.g., studi, histori).
# May over-stem or under-stem words.
# Less accurate than lemmatization.
```

```python
# Snowball Stemmer

# An improved version of Porter Stemmer.
# Supports multiple languages.
# Example:

from nltk.stem import SnowballStemmer

stemmer = SnowballStemmer("english")

print(stemmer.stem("playing"))  
print(stemmer.stem("studies")) 
print(stemmer.stem("fairly"))
print(stemmer.stem("female"))
print(stemmer.stem("male"))
# Snowball Stemmer

# An improved version of Porter Stemmer.
# Supports multiple languages.
# Example:

from nltk.stem import SnowballStemmer

stemmer = SnowballStemmer("english")

print(stemmer.stem("playing"))  
print(stemmer.stem("studies")) 
print(stemmer.stem("fairly"))
print(stemmer.stem("female"))
print(stemmer.stem("male"))
print(stemmer.stem("history"))
```

```python
# The Lancaster Stemmer is a stemming algorithm used in text processing to reduce words to their root form. 
# It is known for being more aggressive than the Porter Stemmer, 
# sometimes producing stems that are not actual dictionary words.

from nltk.stem import LancasterStemmer

stemmer = LancasterStemmer()

words = [
   "playing", "played", "plays", "studies", "studying","eating","eat","eaten","goes","go","female","male","history"
]

for word in words:
    print(f"{word} -> {stemmer.stem(word)}")
```

```python
# A RegexStemmer is a stemmer that uses regular expression (regex) patterns to remove predefined suffixes from words. 
# Unlike the Lancaster Stemmer or Porter Stemmer, it does not rely on linguistic rules.
#  Instead, it strips suffixes that match a regex pattern.

# Example

# Suppose we want to remove common English suffixes: ing, ed, and s.

from nltk.stem import RegexpStemmer

stemmer = RegexpStemmer('ing$|ed$|s$', min=3)

words = ["playing", "played", "plays", "cats", "reading"]

for word in words:
    print(f"{word} -> {stemmer.stem(word)}")
```
