```python
from nltk.tokenize import sent_tokenize
# corpus is nothing but a paragraph
# document is sentences/tokens
# vocubulary is unique words
# words is tokens/whole words

corpus = """Hello Welcome, to learning session.
Today we are going to learn about NLP! Let's start it.
"""
```

```python
# Sentence Tokenization - convert a paragraph (corpus) into sentences
document = sent_tokenize(corpus)
print(document)
type(document)
```

```python
for i in document:
    print(i)
```

```python
from nltk import word_tokenize

word_tokenize(corpus)
```

```python
from nltk import wordpunct_tokenize
for i in document:
    print(wordpunct_tokenize(i))
```

```python
from nltk import TreebankWordTokenizer

# this will take fullstop word and full stop as combained - ex: 'session.'

tokenizer = TreebankWordTokenizer()
print(tokenizer.tokenize(corpus))
```
