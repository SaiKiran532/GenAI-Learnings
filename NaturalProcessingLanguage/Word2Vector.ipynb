Word Embeddings

Word Embeddings are a way of representing words as dense vectors of real numbers so that words with similar meanings have similar vector representations.

Unlike Bag of Words and TF-IDF, which create sparse vectors and don't understand meaning, Word Embeddings capture semantic relationships between words.

For example:

King  -> [0.25, -0.14, 0.83, ...]
Queen -> [0.22, -0.10, 0.80, ...]
Man   -> [0.18, -0.32, 0.75, ...]
Woman -> [0.20, -0.28, 0.78, ...]

Notice that similar words have similar vector values.

Why do we need Word Embeddings?

Consider these sentences:

I love dogs.
I like dogs.

Bag of Words
Word	Sentence 1	Sentence 2
love	1	            0
like	0	            1
dogs	1	            1

Bag of Words treats love and like as completely different words.

Word Embeddings
love → [0.21, 0.45, 0.11]
like → [0.23, 0.43, 0.10]

Since the vectors are similar, the model understands that love and like have related meanings.

Word2Vec

Word2Vec is one of the most popular algorithms for learning word embeddings. It was introduced by researchers at Tomas Mikolov and colleagues.

Instead of counting words, it learns vector representations by predicting words from their context (or vice versa).

Word2Vec has two architectures:

CBOW (Continuous Bag of Words)
Skip-Gram

1. CBOW (Continuous Bag of Words)
Idea

Predict the current (target) word using the surrounding (context) words.

Example

Sentence:

I love learning NLP

Suppose the target word is learning.

Context words:

I love ____ NLP

Model Input:

I
love
NLP

Output:

learning

So,

Context Words → Target Word
Another Example

Sentence:

The cat drinks milk

Input:

The
cat
milk

Output:

drinks
CBOW Diagram
      I
      love
      NLP
        │
        ▼
     Neural Network
        │
        ▼
    learning
Advantages of CBOW
Faster to train.
Works well for frequent words.
Suitable for large datasets.
Disadvantages
Less effective for rare words.
Averages the context, which can lose some information.

2. Skip-Gram

Idea

Predict the surrounding (context) words from the current (target) word.

Sentence:

I love learning NLP

Target word:

learning

Output:

I
love
NLP

So,

Target Word → Context Words
Another Example

Sentence:

The cat drinks milk

Input:

drinks

Output:

The
cat
milk
Skip-Gram Diagram
      learning
           │
           ▼
      Neural Network
           │
           ▼
   I   love   NLP
Advantages of Skip-Gram
Better for rare words.
Produces higher-quality embeddings.
Often performs better on smaller datasets.
Disadvantages
Slower to train than CBOW.
Requires more computation.

Semantic Relationships

One famous property of Word2Vec is that it captures relationships between words.

Example:

King - Man + Woman ≈ Queen

Similarly:

Paris - France + Italy ≈ Rome

This works because the learned vectors encode semantic relationships.



Limitations of Word2Vec
One vector per word
"Bank" (river bank) and "Bank" (financial institution) share the same embedding.
Cannot fully understand long-range context
It mainly relies on nearby words.
Out-of-vocabulary words
Words not seen during training cannot be represented (though models like FastText address this by using subword information).

Interview Questions
1. Why are Word Embeddings better than TF-IDF?
TF-IDF measures word importance but does not understand meaning.
Word embeddings learn semantic similarity, so related words have similar vectors.

2. What is the difference between CBOW and Skip-Gram?
CBOW: Uses surrounding words to predict the missing word.
Skip-Gram: Uses the current word to predict surrounding words.

3. Which is faster?
CBOW is faster.
Skip-Gram is slower but generally produces better embeddings for rare words.

Average Word2Vec (also called Average Word Embeddings) is a simple technique to represent an entire sentence or document using Word2Vec.

Remember:

Word2Vec gives a vector for each word.
But machine learning models often need one vector per sentence/document.
So we simply average all the word vectors.
Example

Suppose Word2Vec has learned 3-dimensional vectors (real models use 100, 200, or 300 dimensions).

Sentence:

I love cricket

Word vectors:

Word	Vector
I	[0.2, 0.1, 0.4]
love	[0.5, 0.6, 0.3]
cricket	[0.8, 0.7, 0.9]
Step 1: Add corresponding dimensions
Dimension 1:
0.2 + 0.5 + 0.8 = 1.5

Dimension 2:
0.1 + 0.6 + 0.7 = 1.4

Dimension 3:
0.4 + 0.3 + 0.9 = 1.6
Step 2: Divide by number of words (3)
Average Vector

[
1.5/3,
1.4/3,
1.6/3
]

=

[0.50, 0.47, 0.53]

This becomes the sentence vector.

Visual Representation
                 Word2Vec

I        ---> [0.2, 0.1, 0.4]

love     ---> [0.5, 0.6, 0.3]

cricket  ---> [0.8, 0.7, 0.9]
                 │
                 ▼
        Take Average
                 │
                 ▼
      [0.50, 0.47, 0.53]

Now the entire sentence is represented by a single vector.

Another Example

Sentence:

Dogs love food

Suppose

Dogs -> [1, 2]

love -> [2, 3]

food -> [3, 4]

Average:

Dimension 1:

(1 + 2 + 3)/3 = 2

Dimension 2:

(2 + 3 + 4)/3 = 3

Sentence Vector

[2, 3]

Interview Question

Q: Why do we need Average Word2Vec?

Answer:

Word2Vec generates embeddings for individual words.
Most machine learning algorithms require a fixed-size feature vector for each sentence or document.
Average Word2Vec combines all the word embeddings by taking their mean, producing a single fixed-length vector regardless of sentence length.

Gensim and GloVe

Many beginners confuse Gensim and GloVe, but they are different things.

Gensim is a Python library used to train and use word embedding models like Word2Vec, FastText, and Doc2Vec.
GloVe (Global Vectors for Word Representation) is a word embedding algorithm/model that learns word vectors using global word co-occurrence statistics.

```python
from gensim.models import Word2Vec

# Training data
sentences = [
    ["i", "love", "cricket"],
    ["i", "love", "football"],
    ["cricket", "is", "my", "favorite", "game"]
]

# Train model
model = Word2Vec(
    sentences,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4
)
```

Explanation of parameters
Parameter	    Meaning
vector_size=100	Embedding size (100 dimensions)
window=5	    Context window size
min_count=1	    Keep all words appearing at least once
workers=4	    Use 4 CPU cores

```python
print(model.wv["cricket"])
vec = model.wv["cricket"]
```

```python
model.wv.most_similar("cricket")
```

```python
import numpy as np

sentence = ["i", "love", "cricket"]

vectors = [model.wv[word] for word in sentence]

sentence_vector = np.mean(vectors, axis=0)

print(sentence_vector.shape)
```
