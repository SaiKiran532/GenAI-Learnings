Lemmatization is the process of reducing a word to its lemma,

which is its base or dictionary form. Unlike stemming, lemmatization uses vocabulary and grammatical 

(part-of-speech) information to return a valid word.

This is slower than Stemming but predicts correct words

used in Q&A , Chatbots & text summarization



```python
import nltk
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

### pos - v--> verb a --> adjective n--> noun

print(lemmatizer.lemmatize("running", pos="v"))   # run
print(lemmatizer.lemmatize("better", pos="a"))    # good
print(lemmatizer.lemmatize("mice"))               # mouse
print(lemmatizer.lemmatize("studies", pos="v"))   # study
```

```python
words = [
   "playing", "played", "plays", "studies", "studying","eating","eat","eaten","goes","go","female","male","history","fairly","sportingly"
]

for word in words:
    print(f"{word} -> {lemmatizer.lemmatize(word,pos='v')}")
```
