POS (Part-of-Speech) Tagging in Natural Language Processing (NLP) is the process of assigning a grammatical label to each word in a sentence based on its role.

Some common POS tags are:

| Word Type   | Example   | POS Tag |
| ----------- | --------- | ------- |
| Noun        | cat, book | NN      |
| Verb        | run, eat  | VB      |
| Adjective   | beautiful | JJ      |
| Adverb      | quickly   | RB      |
| Pronoun     | he, she   | PRP     |
| Preposition | in, on    | IN      |
| Determiner  | the, a    | DT      |
| Conjunction | and, but  | CC      |

Example

Sentence:

"The quick brown fox jumps over the lazy dog."


| Word  | POS Tag | Meaning                    |
| ----- | ------- | -------------------------- |
| The   | DT      | Determiner                 |
| quick | JJ      | Adjective                  |
| brown | JJ      | Adjective                  |
| fox   | NN      | Noun                       |
| jumps | VBZ     | Verb (3rd person singular) |
| over  | IN      | Preposition                |
| the   | DT      | Determiner                 |
| lazy  | JJ      | Adjective                  |
| dog   | NN      | Noun                       |

```python
import nltk
from nltk.tokenize import word_tokenize

sentence = "The quick brown fox jumps over the lazy dog."

# Tokenize the sentence
words = word_tokenize(sentence)
print(words)

# Perform POS tagging
pos_tags = nltk.pos_tag(words)

print(pos_tags)
```

```python
import nltk
from nltk.tokenize import word_tokenize

sentence = "ChatGPT helps students learn NLP."

words = word_tokenize(sentence)
tags = nltk.pos_tag(words)

print("Word\t\tPOS")
print("-" * 25)

for word, tag in tags:
    print(f"{word:<12}{tag}")
```
