### if, else & elif and Nested statement

```python
#if and else statements

age = 20

if age>=18:
    print('You are allowed to vote in the election')
else:
    print('You are not allowed to vote in the election')
```

```python
#elif statement

age = 17

if age<13:
    print('You are Child')
elif age<18 :
    print('You are a teenager')
else:
    print('You are an adult')
```

```python
#nested conditional statement

num = int(input('Enter the number'))

if num>=0:
    print('The number ' + str(num) + ' is positive')
    if num%2==0:
        print('The number is even')
    else:
        print('The number is odd')
else:
    print('The number is zero or negitive')
```

```python
# Leap year program

year = int(input('Enter the year:'))

if year%4==0:
    if year%100==0:
        if year%400==0:
            print(year,'Year is a leap year')
        else:
            print(year,'Year is not a leap year')
    else:
        print(year,'Year is not a leap year')
else:
   print(year,'Year is not a leap year')               
```

### Loops

```python
# break & continue

for i in range(10):
    if i==5:
        break
    print(i)
```

```python
# continue statment skips the current iteration anf continues with the next.
for i in range(10):
    if i%2==0:
        continue
    print(i)
```

```python
# while loop

count = 0

while count<=5:
    print(count)
    count+=1
```

```python
# pass keyword

for i in range(1,10,2):
    if i == 3:
        pass
    print(i)
```

```python
#  Nested loops

for i in range(3):
    for j in range(2):
        print(f'i:{i} and j:{j}')
```

```python
# prime number

for num in range(1,15):
    if num>1:
        for i in range(2,num):
            if num%i==0:
                break
        else:
            print(num)
```

```python
### Nested Dictionaries
students = {
    'student1':{'Name':'Sai','Age':27},
    'student2':{'Name':'Kiran','Age':34},
}

print(students)
print(students['student1']['Name'])
print(students['student2']['Age'])
```

```python
for student_id,student_info in students.items():
    print(f'{student_id} -> {student_info}')
    for key,value in student_info.items():
        print(f'{key} : {value}')
```

```python
# Use a Dic to count the frequency of the numbers in a list

numbers = [1,2,1,2,2,2,3,4,5,5]
freq = {}

for i in numbers:
    if i in freq:
        freq[i]+=1
    else:
        freq[i]=1
print(freq)
```

### Functions

```python
### Keyword arguments & Positional arguments
def keywords_arg(*args, **kwargs):
    for i in args:
        print(f"positional args: {i}")

    for key, value in kwargs.items():
        print(f"{key}: {value}")

keywords_arg(1,2,name='Sai',age='27')
```

### Lambda Arguments

```python
def add(a,b):
    return a+b
print(add(2,3))


addition = lambda a,b: a+b
print(addition(6,8))
```

```python
## map syntax

l = list(map(int,input("Enter n numbers").split()))
print(l)
```

```python
#Filter 

def even(num):
    if num%2==0:
        return True


l = [1,2,3,4,5,6,7,8]

list(filter(even,l))
```

```python
import csv

with open('example.csv',mode='w',newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['name','age'])
    writer.writerow(['Saikiran',27])


with open('example.csv',mode='r',newline='') as file:
    reader = csv.reader(file)
    for i in reader:
        print(i)
```
