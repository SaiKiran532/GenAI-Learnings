```python
class BankAccount:
    def __init__(self, account_holder, balance=0):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited ₹{amount}")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"Withdrawn ₹{amount}")
        else:
            print("Insufficient Balance!")

    def display_balance(self):
        print(f"Account Holder: {self.account_holder}")
        print(f"Current Balance: ₹{self.balance}")


# Creating an object
account1 = BankAccount("Ravi", 1000)

# Using methods
account1.display_balance()
account1.deposit(500)
account1.display_balance()
account1.withdraw(300)
account1.display_balance()
```

### Inheritance

```python
class Person:
    def show_name(self):
        print("Name: Ravi")

class Student(Person):
    def show_course(self):
        print("Course: Python")

# Object of child class
s = Student()

s.show_name()    # Inherited from Person
s.show_course()  # Own method
```

```python
class Father:
    def father_skill(self):
        print("Father: Driving")

class Mother:
    def mother_skill(self):
        print("Mother: Cooking")

class Child(Father, Mother):
    def child_skill(self):
        print("Child: Coding")

# Object of child class
c = Child()

c.father_skill()  # From Father
c.mother_skill()  # From Mother
c.child_skill()   # Own method
```

### Polymorphism

```python
# Parent class 1
class Dog:
    # Same method name in all classes
    def sound(self):
        print("Dog barks")


# Parent class 2
class Cat:
    # Same method name but different behavior
    def sound(self):
        print("Cat meows")


# Parent class 3
class Cow:
    # Same method name but different behavior
    def sound(self):
        print("Cow moos")


# Creating objects of different classes
animals = [Dog(), Cat(), Cow()]

# Polymorphism:
# The same method call (sound()) works differently
# depending on the object type.
for animal in animals:
    animal.sound()
```
